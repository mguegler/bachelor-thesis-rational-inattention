```python
# Setup

import os
import numpy as np
import auswertung as aw
from modell_parameter import ModelParams
from modell_konsument import DynamicRIModel

os.makedirs("abbildungen", exist_ok=True)

baseline = ModelParams()
```

```python
# Berechnungen

grenzfaelle_lambda = [
    (r"$\lambda=0$", DynamicRIModel(ModelParams(lambda_=0.0)).solve_dynamic_model()),
    (r"$\lambda\to\infty$", DynamicRIModel(ModelParams(lambda_=np.inf)).solve_dynamic_model())]
    
grenzfaelle_rho = [
    (r"$\rho=0$", DynamicRIModel(ModelParams(rho=0.0)).solve_dynamic_model()),
    (r"$\rho=0.95$", DynamicRIModel(ModelParams(rho=0.95)).solve_dynamic_model())]

demand = aw.compute_demand(baseline)
```

```python
# Robustheitstabellen

tabellen = aw.robustness_checks(baseline, demand=demand, output_dir="tabellen_anhang")
```

```python
# Abbildungen Grenzfälle

aw.fig_limits(grenzfaelle_lambda).savefig("abbildungen/anhang_grenzfaelle_lambda.pdf", bbox_inches="tight")
aw.fig_limits(grenzfaelle_rho).savefig("abbildungen/anhang_grenzfaelle_rho.pdf", bbox_inches="tight")
```
